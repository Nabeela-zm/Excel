# String Concatenation

```python
str1="Hello"
str2=input("Enter your name :")
str3=str1+","+str2
print(str3)
```

```python
str4=",welocome to Python programming"
str5=str3+str4
print(str5)
```

# string slicing and indexing

```python
print(str5[0])
print(str5[-1])
print(str5[:5:])
print(str5[-11:])
print(str5[::-1])
print(str5[-18:-11])
```

# string methods

```python

strM="Python beginner tutorial"

print(strM.upper())
print(strM.lower())
print(strM.capitalize())
print(strM.count("t"))
strN=strM.replace("Python","Machine learning")
print(strN)

```

# Tuple creation

```python
t1=(10,20,30)
t2=(40,50,60)
t_combine=t1+t2
print(t_combine)
print(t_combine*3)
print(t_combine[2])
print(t_combine[:3])
print(t_combine[-3:])
```
